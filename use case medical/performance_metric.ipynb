```python
import pandas as pd
import matplotlib.pyplot as plt

# Synthetic example data
data = {
    "Metric": ["Avg Time (ms)", "Min Time (ms)", "Max Time (ms)", "Precision@10"],
    "Value": [42, 30, 78, 0.87]
}
df = pd.DataFrame(data)

df
```

```python
import matplotlib.pyplot as plt

# Bar chart for time metrics
times = df.iloc[:3]
plt.figure(figsize=(6,4))
plt.bar(times["Metric"], times["Value"])
plt.xlabel("Metric")
plt.ylabel("Milliseconds")
plt.title("k-NN Query Time Performance (Text)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
