```python
import numpy as np
import oracledb
from sklearn.metrics.pairwise import cosine_similarity

# ======================
# CONFIG ORACLE
# ======================
user = "ryan"
password = "ryan23"
dsn = "localhost:1521/FREEPDB1"

# ======================
# 1. Charger le nouveau patient depuis Oracle
# ======================
# ID du nouveau patient à comparer
nouveau_patient_id = 4

conn = oracledb.connect(user=user, password=password, dsn=dsn)
cur = conn.cursor()

cur.execute("""
    SELECT ID, description, vector
    FROM new_patient
    WHERE id = :id
""", {"id": nouveau_patient_id})

row = cur.fetchone()

if not row:
    raise ValueError(f"⚠ Aucun patient trouvé avec id={nouveau_patient_id}")

nouveau_id, nouveau_desc, nouveau_vec_raw = row
# Convertir le vecteur brut en un tableau numpy
nouveau_vecteur = np.array(nouveau_vec_raw, dtype=np.float32)

print(f"Nouveau patient : ID={nouveau_id}")
print(f"Description : {nouveau_desc}")
print(f"Vecteur dimension : {nouveau_vecteur.shape[0]}")


# ======================
# 2. Charger les patients existants depuis Oracle
# ======================

cur.execute("""
    SELECT id, description, traitement_prescrit, outcome, vector
    FROM patient_cases
""")
rows = cur.fetchall()


patients_data = []
for id_, desc, traitement, outcome, vect in rows:
    try:
        # Conversion directe en array NumPy
        vecteur = np.array(vect, dtype=np.float32)
        patients_data.append((id_, desc, traitement, outcome, vecteur))
    except Exception as e:
        print(f"⚠ Erreur conversion vecteur pour patient ID {id_} : {e}")

# ======================
# 3. Calculer similarité cosinus
# ======================
vectors = [p[4] for p in patients_data]
scores = cosine_similarity([nouveau_vecteur], vectors)[0]

patients_scores = []
for (id_, desc, traitement, outcome, vect), score in zip(patients_data, scores):
    patients_scores.append((id_, desc, traitement, outcome, vect, score))

patients_scores.sort(key=lambda x: x[5], reverse=True)

# ======================
# 4. Afficher les résultats
# ======================

print("\nRésultats de recherche (par similarité cosinus) :\n")
for id_, desc, traitement, outcome, vect, score in patients_scores[:5]:
    print(f"ID: {id_} | Similarité: {score:.4f}")
    print(f"Description: {desc}")
    print(f"Traitement: {traitement} | Outcome: {outcome}")
    print("-" * 60)

cur.close()
conn.close()
```
